# Collections Dashboard Mockup
A fake dataset to explore trends in collections data.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate mock data
data = pd.DataFrame({
    'customer_id': range(1, 101),
    'delinquent_days': [max(0, x) for x in np.random.normal(20, 10, 100)],
    'amount_due': [round(abs(x), 2) for x in np.random.normal(120, 60, 100)]
})

# Visualize
plt.figure(figsize=(8,5))
sns.histplot(data['delinquent_days'], bins=20, kde=True)
plt.title('Delinquency Days Distribution')
plt.xlabel('Days Past Due')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```
