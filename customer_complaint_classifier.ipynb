# Customer Complaint Classifier
A basic example to classify complaint categories based on keywords.

```
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Sample data
complaints = pd.DataFrame({
    'text': [
        'Late payment fee charged unfairly',
        'Unable to access account',
        'Dispute not resolved',
        'App crashes during payment',
        'Fraudulent activity on my account'
    ],
    'category': ['billing', 'technical', 'dispute', 'technical', 'fraud']
})

# Vectorize
vec = CountVectorizer()
X = vec.fit_transform(complaints['text'])

# Model
model = MultinomialNB()
model.fit(X, complaints['category'])

# Predict sample
sample = vec.transform(['Payment not processed'])
print("Predicted category:", model.predict(sample)[0])
```
